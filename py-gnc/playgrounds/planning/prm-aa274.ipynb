# Heuristic Comparison

```python
configs = [
    (AStarHeuristic.l2, 1.0),
    (AStarHeuristic.l1, 1.0),
    (AStarHeuristic.linf, 1.0),
    (AStarHeuristic.l2, 3.0),
]
colors = ["green", "red", "orange", "purple"]

plt.rcParams["figure.figsize"] = [10, 10]
occupancy.plot()
print(f"{'heuristic':>9} | {'w':>4} | {'path length':>11} | {'nodes expanded':>14}")
print("-" * 49)
for (heuristic, weight), color in zip(configs, colors):
    astar_h = AStarGrid(
        (0, width),
        (0, height),
        x_init,
        x_goal,
        occupancy,
        resolution=0.1,
        heuristic=heuristic,
        heuristic_weight=weight,
    )
    if not astar_h.solve():
        print(f"{heuristic:>9} | {weight:>4.1f} | no path found")
        continue
    print(
        f"{heuristic.name:>9} | {weight:>4.1f} | {path_length(astar_h.path):>11.3f} | {len(astar_h.visited):>14d} | {astar_h.cost_to_arrive[astar_h.x_goal]}"
    )
    p = np.asarray(astar_h.path)
    plt.plot(
        p[:, 0],
        p[:, 1],
        color=color,
        linewidth=2,
        label=f"{heuristic.name}, w = {weight}",
    )
plt.legend()
plt.title("A* paths for different heuristics")
plt.show()
```
